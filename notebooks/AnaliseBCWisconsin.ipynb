Analisando dataset


```python
path = r"D:\Projeto AWS\data\wdbc.data"

columns = [
    "id",
    "diagnosis",

    "mean_radius",
    "mean_texture",
    "mean_perimeter",
    "mean_area",
    "mean_smoothness",
    "mean_compactness",
    "mean_concavity",
    "mean_concave_points",
    "mean_symmetry",
    "mean_fractal_dimension",

    "radius_se",
    "texture_se",
    "perimeter_se",
    "area_se",
    "smoothness_se",
    "compactness_se",
    "concavity_se",
    "concave_points_se",
    "symmetry_se",
    "fractal_dimension_se",

    "worst_radius",
    "worst_texture",
    "worst_perimeter",
    "worst_area",
    "worst_smoothness",
    "worst_compactness",
    "worst_concavity",
    "worst_concave_points",
    "worst_symmetry",
    "worst_fractal_dimension"
]

df = pd.read_csv(path,
    header = None,
    names = columns
)

df.head(15)
```

Analise do Raio medio

```python
sns.boxplot(
    data=df,
    x="diagnosis",
    y="mean_radius"
)

plt.title("Distribuição do raio médio por diagnóstico")
plt.xlabel("Diagnóstico")
plt.ylabel("Raio médio")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="mean_radius",
    hue="diagnosis",
    bins=25,
    kde=True,
    element="step"
)

plt.title("Distribuição do raio médio por diagnóstico")
plt.xlabel("Raio médio")
plt.ylabel("Quantidade de casos")
plt.show()
```

Analise Raio x Textura

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="mean_texture",
    y="mean_radius",
    hue="diagnosis"
)

plt.title("Raio médio × Textura média")
plt.xlabel("Textura média")
plt.ylabel("Raio médio")
plt.show()
```

Analise de concatividade X Textura

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="mean_concavity",
    y="mean_texture",
    hue="diagnosis"
)

plt.title("Concavidade média × Textura média")
plt.xlabel("Concavidade média")
plt.ylabel("Textura média")
plt.show()
```

Analise de 3 features juntas

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

cores = df["diagnosis"].map({
    "B": "orange",
    "M": "blue"
})

ax.scatter(
    df["mean_radius"],
    df["mean_texture"],
    df["mean_perimeter"],
    c=cores,
    alpha=0.7
)

ax.set_title("Raio × Textura × Perímetro por diagnóstico")
ax.set_xlabel("Raio médio")
ax.set_ylabel("Textura média")
ax.set_zlabel("Perímetro médio")

plt.show()
```

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

cores = df["diagnosis"].map({
    "B": "orange",
    "M": "blue"
})

ax.scatter(
    df["mean_texture"],
    df["mean_smoothness"],
    df["mean_concavity"],
    c=cores,
    alpha=0.7
)

ax.set_title("Textura × Suavidade × Concavidade")
ax.set_xlabel("Textura média")
ax.set_ylabel("Suavidade média")
ax.set_zlabel("Concavidade média")

plt.show()
```

```python
sns.boxplot(
    data=df,
    x="diagnosis",
    y="mean_concavity"
)

plt.title("Concavidade média por diagnóstico")
plt.xlabel("Diagnóstico")
plt.ylabel("Concavidade média")
plt.show()
```

Analise das features quanto ao diagnostico

```python
features = df.drop(columns=["id", "diagnosis"]).columns

comparacao = df.groupby("diagnosis")[features].mean().T

comparacao["diferenca_absoluta"] = (
    comparacao["M"] - comparacao["B"]
).abs()

comparacao.sort_values(
    "diferenca_absoluta",
    ascending=False
).head(10)
```
