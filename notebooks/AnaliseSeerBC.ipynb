```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Analise do dataset

```python
path = r"D:\Projeto AWS\data\Breast_Cancer.csv"

columns = [
    "Age",
    "Race",
    "Marital_Status",
    "T_Stage",
    "N_Stage",
    "6th_Stage",
    "differentiate",
    "Grade",
    "A_Stage",
    "Tumor_Size",
    "Estrogen_Status",
    "Progesterone_Status",
    "Regional_Node_Examined",
    "Reginol_Node_Positive",
    "Survival_Months",
    "Status",
]

df = pd.read_csv(path,
    header = None,
    names = columns
)

df.head(15)
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="6th_Stage",
    hue="Status",
    bins=25,
    kde=True,
    element="step"
)

plt.title("Distribuição do raio médio por diagnóstico")
plt.xlabel("6th_Stage")
plt.ylabel("Status de vida")
plt.show()
```
